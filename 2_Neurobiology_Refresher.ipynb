### The Answer: An Electrochemical Gradient
A neuron's "battery" isn't a chemical reaction driving electrons through a wire — it's a **concentration difference** of ions (mainly K⁺ and Na⁺) held across the membrane, the same membrane you'll model as a capacitor starting in Set 3. K⁺ is concentrated *inside* the cell; Na⁺ is concentrated *outside*. Separating charge like this, across an insulator, stores potential energy — exactly the way a battery does, just with ion gradients as the "fuel" instead of a chemical reaction.

### But Gradients Leak — So What Recharges Them?
Leak channels (Pillar 3, and the ones you already "bought" in Set 3's Module A) constantly let ions drift down their gradients, and every action potential lets even more through. Left alone, the battery would run down. Something has to constantly recharge it.

### Quantifying the Battery: The Nernst Equation
How much voltage does a given gradient actually store? That's what the **Nernst equation** answers — for a single ion species, it gives the membrane voltage at which that ion's electrical and chemical forces exactly balance:

$$E_{ion} = \frac{RT}{zF} \ln\left(\frac{[\text{ion}]_{outside}}{[\text{ion}]_{inside}}\right)$$

where $R$ is the gas constant, $T$ is temperature (Kelvin), $z$ is the ion's charge (+1 for K⁺), and $F$ is Faraday's constant. Don't worry about memorizing this — you're about to have AI help you turn it into code.

## Pillar 5 — How Circuits Learn: Hebb's Rule
Two traits are hardwired into essentially every animal species, from insects to humans: **survival and reproduction**. One way to survive is *innate* fear — some fears you're simply born with. The other is the ability to **learn fear fast**.

Think about touching a hot stove for the first time. You don't need a lesson, a textbook, or a second try — one bad experience, and it's locked into memory. That's not a coincidence; it reflects a circuit-level rule for how connections between neurons change:

> **Hebb's Rule (Donald O. Hebb, 1949): "Cells that fire together, wire together."**

When two neurons are active at the same time, again and again, the connection between them tends to strengthen. This one idea is the conceptual seed for two things later in this course: the calcium-based plasticity rule you'll work with directly in **Set 8**, and the very circuit you're about to reverse-engineer below.

### The Three Phases
- **Habituation** (Day 1): the tone is played 5 times, alone. The rat is anxious at first, then gets used to it.
- **Conditioning** (Day 1, later): the tone is played again, but now a brief foot shock co-terminates with it — tone and shock happen *together*, 10 times.
- **Testing** (Day 2): tone only, no shock. A rat that has learned the association freezes as soon as the tone starts.

Fear is measured by **freezing** — the more complete the immobility, the stronger the fear. This can even be automated by tracking how long a rat stays motionless.

### The Biology You Need for This Circuit
Fear is linked to firing of neurons in the **amygdala** — specifically its central nucleus, a key site for learning fear in both rodents and humans.

**The tone pathway:** sound is picked up by sensory neurons in the **cochlea** → relayed to the **auditory thalamus** → relayed to the **amygdala**.

**The shock pathway:** the foot shock is picked up by sensory neurons in the paw → relayed to the **pain thalamus** (a different subdivision of the same thalamus) → relayed to the **amygdala**.

**The key point:** both pathways converge on the **same amygdala neuron**. This is exactly the setup Hebb's rule needs — two inputs arriving at one cell, together, repeatedly.
